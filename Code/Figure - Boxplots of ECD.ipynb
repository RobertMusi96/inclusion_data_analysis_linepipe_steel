```python
# Import packages and plotting setup
import pandas as pd
import numpy as np
from pathlib import Path

from matplotlib.ticker import MultipleLocator
import matplotlib.pyplot as plt

from Functions import setup_matplotlib_parameter

wong_cycle = setup_matplotlib_parameter()
```

```python
# Data file imports
data_dir = Path("..") / "Data files" / "Inclusion lists per stage"

df_container = {}
for file in data_dir.glob("*.xlsx"):
    df_container[file.stem] = pd.read_excel(file, sheet_name="NMI data", index_col=0)
```

### Boxplots

```python
fig, axs = plt.subplots(nrows=1, ncols=2,
                        figsize=(210/25.4*0.9, 297/25.4 * 0.3),
                        gridspec_kw={"width_ratios": [1, 1.2]})

all_values = []
labels = []

i = 1
j = 0
for id, df in df_container.items():
    vals = df["ECD (µm)"].values
    all_values.append(vals)
    labels.append(id.split("-")[-1])
    i += 1
    j += 1

    if j == 12 or j == 14 or j == 16 or j == 18:
        i = 1

    if i == 5:
        i = 1

# ----------------------Plot for S1-S3--------------------------
positions = [0, 1, 2, 3,
             5, 6, 7, 8,
             10, 11, 12, 13]

bp = axs[0].boxplot(all_values[0:12], positions=positions,
                    widths=0.8, patch_artist=True, showfliers=False,
                    medianprops=dict(color="black", linewidth=1.5),
                    whiskerprops=dict(color="#333333", linewidth=0.8),
                    capprops=dict(color="#333333", linewidth=0.8),
                    boxprops=dict(edgecolor="#333333", linewidth=0.8))

group_labels = ["Heat-1","Heat-2", "Heat-3", "Heat-4",
                "Heat-1", "Heat-2", "Heat-3", "Heat-4",
                "Heat-1", "Heat-2", "Heat-3", "Heat-4"]

group_colors = {"Heat-1": wong_cycle[0], "Heat-2": wong_cycle[1], "Heat-3": wong_cycle[2], "Heat-4": wong_cycle[3]}
for patch, label in zip(bp['boxes'], group_labels):
    patch.set_edgecolor(group_colors[label])
    patch.set_linewidth(1.0)
    # patch.set_alpha(0.8)
    patch.set_facecolor("white")

# Adjust x axis
axs[0].set_xticks(positions)
axs[0].set_xticklabels(labels[:12])
axs[0].set_xlim(-1, 14)

# Set y axis
axs[0].set_ylabel("ECD [µm]", fontweight="normal", fontsize=12)
axs[0].set_xlabel("Heat", fontweight="normal", fontsize=12)
axs[0].set_ylim(0.4, 5)
axs[0].set_axisbelow(True)

axs[0].set_yticks(np.arange(0.5, 5.1, 0.5))
axs[0].yaxis.set_minor_locator(MultipleLocator(0.25))
axs[0].grid(None)
axs[0].xaxis.grid(False)
axs[0].grid(True, axis="y", which="major", color="#dddddd", linewidth=0.7)
# axs[0].yaxis.grid(True, which="minor", color="#eeeeee", linewidth=0.5)

# Generate boxes for sampling stages
positions = range(len(all_values[0:12]))

stage_groups = {
    "S1-DO": (0, 3),
    "S2-DS": (5, 8),
    "S3-VT": (10, 13),
}

for label, (x0, x1) in stage_groups.items():

    axs[0].text(
        (x0 + x1) / 2,
        1,
        label,
        transform=axs[0].get_xaxis_transform(),
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="normal",
    )

# Stage separator lines
axs[0].axvline(4, color="0.8", lw=0.8)
axs[0].axvline(9, color="0.8", lw=0.8)

# Make whiskers and caps the same color as the boxes
for whisker, label in zip(bp["whiskers"], np.repeat(group_labels, 2)):
    whisker.set_color(group_colors[label])

for cap, label in zip(bp["caps"], np.repeat(group_labels, 2)):
    cap.set_color(group_colors[label])

# ------------------------------Plot for S4-S8-------------------------
positions = [
    0, 1,          # S4
    3, 4,          # S5
    6, 7,          # S6
    9,10,11,12,    # S7
    14,15,16,17    # S8
]

bp = axs[1].boxplot(all_values[12:], positions=positions,
                    widths=0.8, patch_artist=True, showfliers=False,
                    medianprops=dict(color="black", linewidth=1.5),
                    whiskerprops=dict(color="#333333", linewidth=0.8),
                    capprops=dict(color="#333333", linewidth=0.8),
                    boxprops=dict(edgecolor="#333333", linewidth=0.8))

# Coloring
group_labels = ["Heat-1","Heat-2",
                "Heat-1", "Heat-2",
                "Heat-1", "Heat-2",
                "Heat-1", "Heat-2", "Heat-3", "Heat-4",
                "Heat-1", "Heat-2", "Heat-3", "Heat-4"]

group_colors = {"Heat-1": wong_cycle[0], "Heat-2": wong_cycle[1], "Heat-3": wong_cycle[2], "Heat-4": wong_cycle[3]}

for patch, label in zip(bp['boxes'], group_labels):
    patch.set_edgecolor(group_colors[label])
    patch.set_linewidth(1.0)
    # patch.set_alpha(0.8)
    patch.set_facecolor("white")


# Recolor medians that sit on dark box fills so they stay visible
for median, label in zip(bp['medians'], group_labels):
    if label == "S8-TU":
        median.set_color("white")
        median.set_linewidth(0.8)

# Generate boxes for sampling stages
stage_groups = {
    "S4-CT": (0, 1),
    "S5-CT": (3, 4),
    "S6-CT": (6, 7),
    "S7-CT": (9, 12),
    "S8-TU": (14, 17),
}

# Add stage labels
for label, (x0, x1) in stage_groups.items():

    axs[1].text(
        (x0 + x1) / 2,
        1,
        label,
        transform=axs[1].get_xaxis_transform(),
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="normal",
    )

# Stage separator lines
axs[1].axvline(2, color="0.8", lw=0.8)
axs[1].axvline(5, color="0.8", lw=0.8)
axs[1].axvline(8, color="0.8", lw=0.8)
axs[1].axvline(13, color="0.8", lw=0.8)

# Setup x axis
axs[1].set_xticks(positions)
axs[1].set_xticklabels(labels[12:])
axs[1].set_xlim(-1, 18)
axs[1].set_xlabel("Heat", fontweight="normal", fontsize=12)

# Setup y axis
axs[1].set_ylim(0.45, 3)
axs[1].set_yticks(np.arange(0.5, 3.1, 0.5))
axs[1].yaxis.set_minor_locator(MultipleLocator(0.25))
axs[1].xaxis.grid(False)
axs[1].yaxis.grid(True, which="major", color="#e8e8e8", linewidth=0.7)
# axs[1].yaxis.grid(True, which="minor", color="#f3f3f3", linewidth=0.5)

axs[1].set_axisbelow(True)
# axs[1].tick_params(labelleft=False)

# Remove top and right spine
for ax in axs:
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

# Make whiskers and caps the same color as the boxes
for whisker, label in zip(bp["whiskers"], np.repeat(group_labels, 2)):
    whisker.set_color(group_colors[label])

for cap, label in zip(bp["caps"], np.repeat(group_labels, 2)):
    cap.set_color(group_colors[label])

plt.tight_layout()
plt.show()
```
